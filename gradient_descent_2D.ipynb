```python
import numpy as np
import matplotlib.pyplot as plt

# function
def f(x, y):
    return x**2 + y**2

# derivative
def df(x, y):
    return np.array([2*x, 2*y])

# gradient descent parameters
x_start, y_start = 8, -6   # initial guess
learning_rate = 0.1
n_iterations = 20

# for storing steps
steps = [(x_start, y_start, f(x_start, y_start))]

# perform gradient descent
x_current, y_current = x_start, y_start
for _ in range(n_iterations):
    grad = df(x_current, y_current)
    x_current -= learning_rate * grad[0]
    y_current -= learning_rate * grad[1]
    steps.append((x_current, y_current, f(x_current, y_current)))

# convert steps to arrays for plotting
steps = np.array(steps)

# plotting the surface
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

# meshgrid for surface
X = np.linspace(-10, 10, 100)
Y = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(X, Y)
Z = f(X, Y)

# plot the surface
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.6)

# plot the descent path
ax.plot(steps[:, 0], steps[:, 1], steps[:, 2], marker='o', color='red', label="Descent Path")
ax.scatter(steps[:, 0], steps[:, 1], steps[:, 2], color='red')

# labels
ax.set_title("Gradient Descent on f(x, y) = x² + y²")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x, y)")
ax.legend()
plt.show()
```
